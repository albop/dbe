```python
# import dataset
import vega_datasets
df = vega_datasets.data('iris')
df
```

```python
df.columns = ['sLength', 'sWidth', 'pLength', 'pWidth', 'species']
```

```python
df.head() ## calling ()
```

```python
df['sLength']   #index with []
```

```python
df['totalLength'] = 0.5*df['sLength'] + 0.5*df['pLength']
```

```python
df.head()
```

```python
df['species'].unique()
```

```python
# extract all virginica lines
mask = df['species'] == 'virginica'  ### '==' : check for equality > True or False
```
