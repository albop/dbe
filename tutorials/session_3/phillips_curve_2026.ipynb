### Method 2: Using groupby

The groupby method provides a more efficient and elegant approach to compute statistics by groups. 

Instead of manually filtering data for each country (as in Method 1), groupby automatically splits the DataFrame into separate groups based on the values in a specified column (like 'Country'), then applies statistical functions (mean, std, etc.) to each group independently. This eliminates the need for loops and makes the code more concise and readable.

This approach is particularly useful when you have many groups or when you want to apply multiple statistical functions at once to the grouped data.

## Step 4: Merging the Databases

__The following command merges the two databases together. Explain the role of argument `on`. What happened to the column names?__

```python
# Merge (join) the two tables together based on common columns
# on=["period", 'Country'] means: match rows that have the same date AND same country
# This combines inflation and unemployment data for each country-date combination
table = table_1.merge(table_2, on=["period", 'Country']) 

table.columns.unique()  # check that we have no duplicate column names
```

## Step 7: Visualizing the Regression

The Phillips Curve suggests a negative relationship between unemployment and inflation. To test this hypothesis statistically, we'll perform a linear regression where inflation is the dependent variable (y) and unemployment is the independent variable (x).

**Regression equation:** `inflation = α + β × unemployment + ε`

Where:
- **α** (alpha) = intercept (expected inflation when unemployment is zero)
- **β** (beta) = slope coefficient (how much inflation changes for each 1% change in unemployment)
- **ε** (epsilon) = error term (unexplained variation)

If the Phillips Curve holds, we expect **β to be negative** (higher unemployment → lower inflation).

## Step 8: Comparing All Countries

So far we've analyzed France in detail and found strong evidence for the Phillips Curve. But does this relationship hold across different economies?

**First approach: Pooled regression**
Let's start by running a single regression using data from all countries combined. This gives us an "average" Phillips Curve across all four economies.

```python
# Run a pooled regression using all countries' data combined
# This treats all observations equally, regardless of country
model_pooled = sm.ols(formula='inflation ~ unemployment', data=df)
result_pooled = model_pooled.fit()

print("Pooled Regression Results (All Countries Combined):")
print(result_pooled.summary())
```

**Visualizing the pooled relationship:**

Now let's plot all countries together with the pooled regression line as a reference. This helps us see:
- Whether the overall negative relationship holds
- How much individual countries deviate from the average pattern

```python
# Create a plot showing all countries with the pooled regression line
# First, generate predictions from the pooled model
df['pooled_pred'] = result_pooled.predict(df['unemployment'])

# Plot all observations
plt.plot(df['unemployment'], df['inflation'], '.')

# Add the pooled regression line 
plt.plot(df['unemployment'], df['pooled_pred'], '-')
```

Now lets make this look nicer

```python
# Create a plot showing all countries with the pooled regression line
# First, generate predictions from the pooled model
df['pooled_pred'] = result_pooled.predict(df['unemployment'])

# Plot all countries with different colors
for country in df['country'].unique():
    df_country = df[df['country'] == country]
    plt.plot(df_country['unemployment'], df_country['inflation'], '.',  label=country)

# Add the pooled regression line (common for all countries)
plt.plot(df['unemployment'], df['pooled_pred'], '-',  label='Pooled regression')

plt.xlabel("Unemployment (%)")
plt.ylabel("Inflation (%)")
plt.title("Phillips Curve - All Countries with Pooled Regression")
plt.legend()
plt.grid()
```

__Now let's go further: Can you find a way to create separate regression lines for each country (all on one graph, or using subplots)?__

**Hint** Use loops to automate the analysis for each country individually, avoiding repetitive code.

```python
# Get a list of all unique countries in our dataset to loop through
countries_list = df['country'].unique()
countries_list
```

```python
# Loop through each country in our dataset


for country in countries_list:
    
    # Filter data for current country using boolean indexing
    df_country = df[df['country'] == country].copy() # boelan indexing (Note: Try running without using .copy() and see what happens when )
    # df_country = df.query(f"country=='{country}'").copy() # alternatively, query syntax
    
    # Create and fit the linear regression model
    model = sm.ols(formula='inflation ~ unemployment', data=df_country)
    result = model.fit()  # Estimate the coefficients
    
    # Generate predicted inflation values from the model
    # Store them in a new column for plotting the regression line
    df_country['reg_inflation'] = result.predict(df_country['unemployment'])
    
    # Plot actual data points as dots
    plt.plot(df_country['unemployment'], df_country['inflation'], '.', 
             label=country, alpha=0.5)
    
    # Plot the regression line (predicted values)
    plt.plot(df_country['unemployment'], df_country['reg_inflation'])
    
    # Add labels and formatting
    plt.title("Phillips Curve")
    plt.xlabel("Unemployment (%)")
    plt.ylabel("Inflation (%)")
    plt.legend()
    plt.grid()
```

Maybe nicer with subplots instead of plotting all countries on the same graph, we can create separate subplots for each country. This allows us to see the individual Phillips curve relationship for each country without overlap.


```python
# Create a 2x2 grid of subplots to later compare all 4 countries side by side
plt.subplot(2,2,1)
plt.subplot(2,2,2)
plt.subplot(2,2,3)
plt.subplot(2,2,4)
```

```python

# enumerate() gives us both the index (i) and the country name in each loop iteration

for i, country in enumerate(countries_list):
    
    # Filter data for the current country using boolean indexing
    df_country = df[df['country'] == country].copy()
    #df_country = df.query(f"country=='{country}'").copy() # alternatively, query syntax

    
    # Run regression for this country
    model = sm.ols(formula='inflation ~ unemployment', data=df_country)
    result = model.fit()
    
    # Generate predicted values for the regression line
    df_country['reg_inflation'] = result.predict(df_country['unemployment'])
    
    # Select subplot position: i+1 because subplots are numbered 1,2,3,4
    plt.subplot(2, 2, i+1)
    
    # Plot actual data points (small dots, semi-transparent)
    plt.plot(df_country['unemployment'], df_country['inflation'], '.', 
             label=country, alpha=0.5)
    
    # Plot regression line
    plt.plot(df_country['unemployment'], df_country['reg_inflation'])
    
    # Add labels and formatting to each subplot
    plt.xlabel("Unemployment (%)")
    plt.ylabel("Inflation (%)")
    plt.legend()
    plt.grid()

# Adjust spacing between subplots for better readability
plt.tight_layout()
```
